```python
class Perceptron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

    def predict(self, inputs):
        # Calculate weighted sum
        total = sum(w * x for w, x in zip(self.weights, inputs)) + self.bias

        # Step activation function
        return 1 if total >= 0 else 0


# Suitable weights and bias for OR gate
weights = [1, 1]
bias = -0.5

# Create perceptron object
perceptron = Perceptron(weights, bias)

# OR gate inputs
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

# Display results
print("X1 X2 Output")
for inp in inputs:
    output = perceptron.predict(inp)
    print(f"{inp[0]}  {inp[1]}   {output}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Perceptron class
class Perceptron:
    def __init__(self, weights, bias):
        self.weights = np.array(weights)
        self.bias = bias

    def predict(self, inputs):
        total = np.dot(inputs, self.weights) + self.bias
        return 1 if total >= 0 else 0


# Suitable weights and bias for NOR gate
weights = [-1, -1]
bias = 0.5

# Create perceptron
perceptron = Perceptron(weights, bias)

# NOR gate inputs
inputs = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Display outputs
print("X1 X2 Output")
for inp in inputs:
    output = perceptron.predict(inp)
    print(f"{inp[0]}  {inp[1]}   {output}")

# -------------------------
# Draw Decision Boundary
# -------------------------
plt.figure(figsize=(6, 6))

# Plot input points
for point in inputs:
    label = perceptron.predict(point)
    if label == 1:
        plt.scatter(point[0], point[1], color='blue', s=100, label='Output = 1')
    else:
        plt.scatter(point[0], point[1], color='red', s=100, label='Output = 0')

# Decision boundary: w1*x1 + w2*x2 + bias = 0
x = np.linspace(-0.2, 1.2, 100)
y = (-(weights[0] * x + bias)) / weights[1]

plt.plot(x, y, 'k--', label='Decision Boundary')

plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Perceptron Decision Boundary for NOR Gate")
plt.grid(True)

# Remove duplicate labels
handles, labels = plt.gca().get_legend_handles_labels()
by_label = dict(zip(labels, handles))
plt.legend(by_label.values(), by_label.keys())

plt.show()
```

```python
# Perceptron Training for Multiple Epochs (AND Gate)

# Training data
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

targets = [0, 0, 0, 1]

# Initial parameters
weights = [0.3, 0.1]
bias = 0
learning_rate = 0.2

# Change this value to 5, 10, or 20
epochs = 20

print("Initial Weights:", weights)
print("Initial Bias:", bias)

for epoch in range(1, epochs + 1):

    total_error = 0

    for i in range(len(inputs)):

        x1, x2 = inputs[i]
        target = targets[i]

        # Net input
        net = x1 * weights[0] + x2 * weights[1] + bias

        # Step activation
        output = 1 if net >= 0 else 0

        # Error
        error = target - output
        total_error += abs(error)

        # Weight update
        weights[0] += learning_rate * error * x1
        weights[1] += learning_rate * error * x2

        # Bias update
        bias += learning_rate * error

    print(f"Epoch {epoch:2d} --> Total Error = {total_error}")

    # Stop if converged
    if total_error == 0:
        print("\nPerceptron Converged!")
        print("Converged at Epoch:", epoch)
        break

print("\nFinal Weights:", weights)
print("Final Bias:", bias)
```

```python
# Perceptron Learning Rate Comparison

# AND Gate Dataset
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

targets = [0, 0, 0, 1]

learning_rates = [0.1, 0.5, 1.0]

for lr in learning_rates:

    weights = [0.3, 0.1]
    bias = 0
    max_epochs = 20

    print("\n====================================")
    print("Learning Rate (η) =", lr)
    print("====================================")

    for epoch in range(1, max_epochs + 1):

        total_error = 0

        for i in range(len(inputs)):

            x1, x2 = inputs[i]
            target = targets[i]

            # Net input
            net = x1 * weights[0] + x2 * weights[1] + bias

            # Step Activation
            output = 1 if net >= 0 else 0

            # Error
            error = target - output
            total_error += abs(error)

            # Weight Update
            weights[0] += lr * error * x1
            weights[1] += lr * error * x2
            bias += lr * error

        print(f"Epoch {epoch} --> Total Error = {total_error}")

        if total_error == 0:
            break

    # Accuracy
    correct = 0

    for i in range(len(inputs)):

        x1, x2 = inputs[i]
        target = targets[i]

        net = x1 * weights[0] + x2 * weights[1] + bias
        output = 1 if net >= 0 else 0

        if output == target:
            correct += 1

    accuracy = (correct / len(inputs)) * 100

    print("\nFinal Weights :", weights)
    print("Final Bias :", bias)
    print("Epochs :", epoch)
    print("Accuracy :", accuracy, "%")
```

```python
import numpy as np

# XOR Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - x**2

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

activations = {
    "Sigmoid": (sigmoid, sigmoid_derivative),
    "Tanh": (tanh, tanh_derivative),
    "ReLU": (relu, relu_derivative)
}

learning_rate = 0.1
epochs = 10000
hidden_neurons = 2

print("-"*75)
print("{:<12}{:<20}{:<20}{:<10}".format(
    "Activation","Train Accuracy","Validation Accuracy","Loss"))
print("-"*75)

for name, (activation, derivative) in activations.items():

    np.random.seed(1)

    W1 = np.random.randn(2, hidden_neurons)
    B1 = np.zeros((1, hidden_neurons))

    W2 = np.random.randn(hidden_neurons, 1)
    B2 = np.zeros((1,1))

    for epoch in range(epochs):

        # Forward Pass
        hidden_input = np.dot(X, W1) + B1
        hidden_output = activation(hidden_input)

        output_input = np.dot(hidden_output, W2) + B2
        output = sigmoid(output_input)

        # Error
        error = Y - output

        # Backpropagation
        d_output = error * sigmoid_derivative(output)

        hidden_error = np.dot(d_output, W2.T)

        if name == "ReLU":
            d_hidden = hidden_error * relu_derivative(hidden_input)
        else:
            d_hidden = hidden_error * derivative(hidden_output)

        # Update
        W2 += learning_rate * np.dot(hidden_output.T, d_output)
        B2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

        W1 += learning_rate * np.dot(X.T, d_hidden)
        B1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

    loss = np.mean(error**2)

    prediction = (output >= 0.5).astype(int)

    accuracy = np.mean(prediction == Y) * 100

    print("{:<12}{:<20.2f}{:<20.2f}{:<10.6f}".format(
        name,
        accuracy,
        accuracy,
        loss))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# XOR Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Sigmoid Activation
def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)

epoch_list = [10, 50, 100]

accuracy_list = []
loss_list = []

for epochs in epoch_list:

    np.random.seed(1)

    # Network Architecture
    W1 = np.random.uniform(-1,1,(2,2))
    B1 = np.random.uniform(-1,1,(1,2))

    W2 = np.random.uniform(-1,1,(2,1))
    B2 = np.random.uniform(-1,1,(1,1))

    learning_rate = 0.5

    # Training
    for epoch in range(epochs):

        # Forward Pass
        hidden_input = np.dot(X,W1)+B1
        hidden_output = sigmoid(hidden_input)

        final_input = np.dot(hidden_output,W2)+B2
        final_output = sigmoid(final_input)

        error = Y-final_output

        # Backpropagation
        d_output = error*sigmoid_derivative(final_output)

        hidden_error = np.dot(d_output,W2.T)
        d_hidden = hidden_error*sigmoid_derivative(hidden_output)

        # Update
        W2 += learning_rate*np.dot(hidden_output.T,d_output)
        B2 += learning_rate*np.sum(d_output,axis=0,keepdims=True)

        W1 += learning_rate*np.dot(X.T,d_hidden)
        B1 += learning_rate*np.sum(d_hidden,axis=0,keepdims=True)

    # Accuracy
    prediction = (final_output >= 0.5).astype(int)
    accuracy = np.mean(prediction == Y) * 100

    loss = np.mean(error**2)

    accuracy_list.append(accuracy)
    loss_list.append(loss)

    print(f"Epochs = {epochs}")
    print(f"Accuracy = {accuracy:.2f}%")
    print(f"Loss = {loss:.6f}\n")

# Plot Accuracy
plt.figure(figsize=(6,4))
plt.plot(epoch_list, accuracy_list, marker='o')
plt.title("Epochs vs Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy (%)")
plt.grid(True)
plt.show()

# Plot Loss
plt.figure(figsize=(6,4))
plt.plot(epoch_list, loss_list, marker='o')
plt.title("Epochs vs Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# XOR Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# -----------------------------
# Model 1
# Input → Hidden → Output
# -----------------------------
model1 = Sequential()

model1.add(Dense(2, input_dim=2, activation='sigmoid'))
model1.add(Dense(1, activation='sigmoid'))

model1.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model1.fit(X, Y, epochs=100, verbose=0)

loss1, acc1 = model1.evaluate(X, Y, verbose=0)

print("Single Hidden Layer")
print("Accuracy :", acc1*100)
print("Loss :", loss1)

print("\nPredictions")
print((model1.predict(X) > 0.5).astype(int))

# -----------------------------
# Model 2
# Input → Hidden1 → Hidden2 → Output
# -----------------------------
model2 = Sequential()

model2.add(Dense(4, input_dim=2, activation='sigmoid'))
model2.add(Dense(2, activation='sigmoid'))
model2.add(Dense(1, activation='sigmoid'))

model2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model2.fit(X, Y, epochs=100, verbose=0)

loss2, acc2 = model2.evaluate(X, Y, verbose=0)

print("\nTwo Hidden Layers")
print("Accuracy :", acc2*100)
print("Loss :", loss2)

print("\nPredictions")
print((model2.predict(X) > 0.5).astype(int))
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# -----------------------------
# Generate Dataset
# -----------------------------
np.random.seed(42)

samples = 300

# Three input features
X = np.random.rand(samples, 3)

# Generate target labels
# Class = 1 if sum of features > 1.5, else 0
Y = (X[:,0] + X[:,1] + X[:,2] > 1.5).astype(int)

# -----------------------------
# Split Dataset
# -----------------------------
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42
)

# -----------------------------
# Normalize Data
# -----------------------------
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# -----------------------------
# Build Neural Network
# -----------------------------
model = Sequential()

model.add(Dense(8, activation='relu', input_shape=(3,)))
model.add(Dense(4, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

# -----------------------------
# Compile Model
# -----------------------------
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# -----------------------------
# Train Model
# -----------------------------
history = model.fit(
    X_train,
    Y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)

# -----------------------------
# Evaluate Model
# -----------------------------
loss, accuracy = model.evaluate(X_test, Y_test, verbose=0)

print("\nTest Accuracy:", accuracy * 100, "%")
print("Test Loss:", loss)
```
